```python
N = 10000         # Definira se broj bacanja kockice

# Generiše se N bacanja kockice
kockica1 = np.random.randint(1, 7, size = N)
kockica2 = np.random.randint(1, 7, size = N)

zbir_brojeva = kockica1 + kockica2

broj_petica = np.sum(zbir_brojeva == 5)

vjerovatnoca = broj_petica/N

print(f"Vjerovatnoća da je suma dobijenih cifara 5 je: {vjerovatnoca * 100:.2f}%.")
```

```python
M = 10000          # Broj ponavljanja eksperimenta
vektor_rezultata = np.zeros(100)

for i in range(M):
    # Baca se novčić dok ne padne pismo
    brojac = 0
    ishodi = ['pismo', 'glava']
    ishod = ''

    while (ishod != 'pismo'):
        # Bacamo novčić
        ishod = np.random.choice(ishodi) 
        brojac += 1

    # U ovom eksperimentu je novčić pao na pismo u "brojač" bacanju
    vektor_rezultata[brojac] += 1

# Ispis prvih 5 vjerovatnoća u postocima
vjerovatnoce_5 = vektor_rezultata[0:5]/M*100
print("Vjerovatnoća da je novčić pao na pismo prvih 5 bacanja:")
print("[" + ", ".join([f"{vj:.2f}%" for vj in vjerovatnoce_5]) + "]")
```

```python
N = 80 # Definira se maksimalan broj ljudi na rođendanu

# Za zadani broj ljudi ponavlja se eksperiment M puta 
# kako bi se "izmjerila" vjerovatnoća
M = 2000

vjerovatnoce = np.zeros(N)

vektor_n = range(N)

for n in vektor_n:    
    broj_pozitivnih = 0
    for m in range(M):        
        # Generišu se dani u godini kada slave rodjendan
        rodjendani = np.random.randint(1, 365+1, size = n)

        # Određuje se da li imaju dva ista broja
        if len(rodjendani) != len(set(rodjendani)):
            # u vektoru postoje dva ista broja
            broj_pozitivnih = broj_pozitivnih + 1
    
    vjerovatnoce[n] = broj_pozitivnih/M


ssp.slprom1D.grafik(vektor_n, vjerovatnoce,
                    x_labela = "$n$", 
                    naslov = "Raspodjela vjerovatnoće $P(n)$")
```
