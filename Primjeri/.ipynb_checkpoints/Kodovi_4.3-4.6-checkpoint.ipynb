```python
# Rucno importovanje ssp modula:
import sys
sys.path.insert(0, '/home/emir/Faks/SIS2/Knjiga_SIS2/Prilozi/SIS2_Kodovi/')
import ssp

# Importovanje ostalih biblioteka:
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import hist

# Podesavanje prikaza za knjigu:
ssp.FORMATIRANJE_PRIKAZA = "BOOK"
ssp.REZOLUCIJA = "LOW"

# Podesavanje prikaza za rad studenata:
# ssp.FORMATIRANJE_PRIKAZA = "JUPYTER"
# ssp.REZOLUCIJA = "LOW"
```

```python
#Programski kod 4.3
```

```python
M = 500 # Broj realizacija slucajnog procesa
N = 500 # Broj uzoraka u jednom slucajnom procesu

# Generisanje vremenske baze
t = np.arange(0, 10, 10/N)


X_t = np.empty((M,N))
a = np.random.uniform(-np.pi, np.pi, M)
A = 1; w = 5;

# PROCES
for k in range(M):
    X_t[k,:] = A * np.cos(w*t + a[k])
    
# Primjeri realizacija
ssp.slproc.realizacije(t, X_t, indeksi = [5,10],
                          naslov = "Prikaz nekoliko realizacija sl.procesa")
```

```python
#Programski kod 4.4
```

```python
# Određivanje očekivane vrijednosti procesa:
eta_X_t = np.empty((N))

for t_i in range(N):
    eta_X_t[t_i] = np.mean(X_t[:,t_i])
    
# Crtanje očekivane vrijednosti procesa eta_X(t)
ssp.slprom1D.grafik(t, eta_X_t, naslov = "$\\eta_x(t)$")
```

```python
#Programski kod 4.5
```

```python
#Određivanje korelacionih funkcija:
R_xx = np.empty((N, N))
C_xx = np.empty((N, N))
for t1 in range(N):
    for t2 in range(N):
        R_xx[t1, t2] = np.mean(X_t[:,t1]*X_t[:,t2])
        C_xx[t1, t2] = np.mean((X_t[:,t1]-eta_X_t[t1])*(X_t[:,t2]-eta_X_t[t2]))
```

```python
#Programski kod 4.6
```
